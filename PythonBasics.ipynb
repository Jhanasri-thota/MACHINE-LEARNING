```python
# ============================================================
# PART 4 - PYTHON SETS
# ============================================================

# ------------------------------------------------------------
# Topic: Create a Set & Basic Operations
# ------------------------------------------------------------
print("\n========== Topic: Create a Set & Basic Operations ==========")

my_set = {1, 2, 3, 4, 5}

print("Original Set:", my_set)

my_set.add(6)
print("After add():", my_set)

my_set.remove(3)
print("After remove():", my_set)

my_set.discard(10)

print("Is 4 present?", 4 in my_set)


# ------------------------------------------------------------
# Topic: Union, Intersection & Difference
# ------------------------------------------------------------
print("\n========== Topic: Union, Intersection & Difference ==========")

A = {1, 2, 3, 4}
B = {3, 4, 5, 6}

print("Union:", A | B)
print("Intersection:", A & B)
print("Difference:", A - B)
print("Symmetric Difference:", A ^ B)


# ------------------------------------------------------------
# Topic: Remove Duplicates from List
# ------------------------------------------------------------
print("\n========== Topic: Remove Duplicates from List ==========")

numbers = [1, 2, 2, 3, 4, 4, 5]

unique_numbers = set(numbers)

print(unique_numbers)


# ------------------------------------------------------------
# Topic: Subset and Superset
# ------------------------------------------------------------
print("\n========== Topic: Subset and Superset ==========")

A = {1, 2, 3}
B = {1, 2, 3, 4, 5}

print("A is subset of B:", A.issubset(B))
print("B is superset of A:", B.issuperset(A))


# ------------------------------------------------------------
# Topic: Set Operations on Strings
# ------------------------------------------------------------
print("\n========== Topic: Set Operations on Strings ==========")

s1 = "hello"
s2 = "world"

set1 = set(s1)
set2 = set(s2)

print("Union:", set1 | set2)
print("Intersection:", set1 & set2)
print("Difference:", set1 - set2)


# ------------------------------------------------------------
# Topic: Find Duplicate Elements
# ------------------------------------------------------------
print("\n========== Topic: Find Duplicate Elements ==========")

def find_duplicates(lst):
    seen = set()
    duplicates = set()

    for item in lst:
        if item in seen:
            duplicates.add(item)
        else:
            seen.add(item)

    return duplicates

numbers = [1, 2, 3, 4, 2, 3, 5, 6, 1]

print(find_duplicates(numbers))


# ------------------------------------------------------------
# Topic: Remove Duplicates while Maintaining Order
# ------------------------------------------------------------
print("\n========== Topic: Remove Duplicates while Maintaining Order ==========")

def remove_duplicates(lst):
    return list(dict.fromkeys(lst))

numbers = [1, 2, 2, 3, 4, 4, 5]

print(remove_duplicates(numbers))


# ------------------------------------------------------------
# Topic: Set Comprehension
# ------------------------------------------------------------
print("\n========== Topic: Set Comprehension ==========")

squared_set = {x ** 2 for x in range(1, 6)}

print(squared_set)


# ------------------------------------------------------------
# Topic: Frozen Set
# ------------------------------------------------------------
print("\n========== Topic: Frozen Set ==========")

A = frozenset([1, 2, 3, 4])

print(A)


# ------------------------------------------------------------
# Topic: Count Unique Words
# ------------------------------------------------------------
print("\n========== Topic: Count Unique Words ==========")

sentence = "python is fun and python is powerful"

words = sentence.split()

unique_words = set(words)

print("Unique Words:", unique_words)
print("Count:", len(unique_words))


# ------------------------------------------------------------
# Topic: Set Methods
# ------------------------------------------------------------
print("\n========== Topic: Set Methods ==========")

sample = {10, 20, 30}

print("Original:", sample)

sample.add(40)
print("After add():", sample)

sample.discard(20)
print("After discard():", sample)

removed = sample.pop()
print("Removed Element:", removed)
print("Remaining Set:", sample)


```

```python
# ============================================================
# PART 7 - SEARCHING ALGORITHMS
# ============================================================

# ------------------------------------------------------------
# Topic: Linear Search
# ------------------------------------------------------------
print("\n========== Topic: Linear Search ==========")

def linear_search(arr, target):
    for index in range(len(arr)):
        if arr[index] == target:
            return index
    return -1

arr = [10, 25, 39, 47, 56]
target = 39

result = linear_search(arr, target)

if result != -1:
    print("Element found at index:", result)
else:
    print("Element not found")


# ------------------------------------------------------------
# Topic: Recursive Linear Search
# ------------------------------------------------------------
print("\n========== Topic: Recursive Linear Search ==========")

def recursive_linear_search(arr, target, index=0):

    if index >= len(arr):
        return -1

    if arr[index] == target:
        return index

    return recursive_linear_search(arr, target, index + 1)

arr = [10, 25, 39, 47, 56]
target = 39

result = recursive_linear_search(arr, target)

if result != -1:
    print("Element found at index:", result)
else:
    print("Element not found")


# ------------------------------------------------------------
# Topic: Binary Search (Iterative)
# ------------------------------------------------------------
print("\n========== Topic: Binary Search (Iterative) ==========")

def binary_search(arr, target):

    left = 0
    right = len(arr) - 1

    while left <= right:

        mid = (left + right) // 2

        if arr[mid] == target:
            return mid

        elif arr[mid] < target:
            left = mid + 1

        else:
            right = mid - 1

    return -1

arr = [10, 20, 30, 40, 50]
target = 30

result = binary_search(arr, target)

if result != -1:
    print("Element found at index:", result)
else:
    print("Element not found")


# ------------------------------------------------------------
# Topic: Recursive Binary Search
# ------------------------------------------------------------
print("\n========== Topic: Recursive Binary Search ==========")

def recursive_binary_search(arr, target, left, right):

    if left > right:
        return -1

    mid = (left + right) // 2

    if arr[mid] == target:
        return mid

    elif arr[mid] < target:
        return recursive_binary_search(arr, target, mid + 1, right)

    else:
        return recursive_binary_search(arr, target, left, mid - 1)

arr = [10, 20, 30, 40, 50]
target = 30

result = recursive_binary_search(arr, target, 0, len(arr) - 1)

if result != -1:
    print("Element found at index:", result)
else:
    print("Element not found")


# ------------------------------------------------------------
# Topic: Compare Linear Search and Binary Search
# ------------------------------------------------------------
print("\n========== Topic: Search Comparison ==========")

print("Linear Search")
print("Best Case    : O(1)")
print("Average Case : O(n)")
print("Worst Case   : O(n)")

print()

print("Binary Search")
print("Best Case    : O(1)")
print("Average Case : O(log n)")
print("Worst Case   : O(log n)")


```

```python
# ============================================================
# PART 8 - SORTING ALGORITHMS
# ============================================================

# ------------------------------------------------------------
# Topic: Bubble Sort
# ------------------------------------------------------------
print("\n========== Topic: Bubble Sort ==========")

def bubble_sort(arr):
    n = len(arr)

    for i in range(n):
        for j in range(0, n - i - 1):

            if arr[j] > arr[j + 1]:
                arr[j], arr[j + 1] = arr[j + 1], arr[j]

numbers = [64, 34, 25, 12, 22, 11, 90]

print("Before Sorting :", numbers)

bubble_sort(numbers)

print("After Sorting  :", numbers)


# ------------------------------------------------------------
# Topic: Selection Sort
# ------------------------------------------------------------
print("\n========== Topic: Selection Sort ==========")

def selection_sort(arr):

    n = len(arr)

    for i in range(n):

        min_index = i

        for j in range(i + 1, n):

            if arr[j] < arr[min_index]:
                min_index = j

        arr[i], arr[min_index] = arr[min_index], arr[i]

numbers = [64, 25, 12, 22, 11]

print("Before Sorting :", numbers)

selection_sort(numbers)

print("After Sorting  :", numbers)


# ------------------------------------------------------------
# Topic: Insertion Sort
# ------------------------------------------------------------
print("\n========== Topic: Insertion Sort ==========")

def insertion_sort(arr):

    for i in range(1, len(arr)):

        key = arr[i]
        j = i - 1

        while j >= 0 and key < arr[j]:

            arr[j + 1] = arr[j]
            j -= 1

        arr[j + 1] = key

numbers = [12, 11, 13, 5, 6]

print("Before Sorting :", numbers)

insertion_sort(numbers)

print("After Sorting  :", numbers)


# ------------------------------------------------------------
# Topic: Merge Sort
# ------------------------------------------------------------
print("\n========== Topic: Merge Sort ==========")

def merge_sort(arr):

    if len(arr) > 1:

        mid = len(arr) // 2

        left = arr[:mid]
        right = arr[mid:]

        merge_sort(left)
        merge_sort(right)

        i = j = k = 0

        while i < len(left) and j < len(right):

            if left[i] < right[j]:
                arr[k] = left[i]
                i += 1

            else:
                arr[k] = right[j]
                j += 1

            k += 1

        while i < len(left):
            arr[k] = left[i]
            i += 1
            k += 1

        while j < len(right):
            arr[k] = right[j]
            j += 1
            k += 1

numbers = [38, 27, 43, 3, 9, 82, 10]

print("Before Sorting :", numbers)

merge_sort(numbers)

print("After Sorting  :", numbers)


# ------------------------------------------------------------
# Topic: Quick Sort
# ------------------------------------------------------------
print("\n========== Topic: Quick Sort ==========")

def quick_sort(arr):

    if len(arr) <= 1:
        return arr

    pivot = arr[len(arr) // 2]

    left = [x for x in arr if x < pivot]
    middle = [x for x in arr if x == pivot]
    right = [x for x in arr if x > pivot]

    return quick_sort(left) + middle + quick_sort(right)

numbers = [10, 7, 8, 9, 1, 5]

print("Before Sorting :", numbers)

sorted_numbers = quick_sort(numbers)

print("After Sorting  :", sorted_numbers)


# ------------------------------------------------------------
# Topic: Built-in sorted() Function
# ------------------------------------------------------------
print("\n========== Topic: Built-in sorted() Function ==========")

numbers = [9, 4, 6, 2, 8]

print("Original :", numbers)

print("Ascending :", sorted(numbers))
print("Descending :", sorted(numbers, reverse=True))


# ------------------------------------------------------------
# Topic: List sort() Method
# ------------------------------------------------------------
print("\n========== Topic: List sort() Method ==========")

numbers = [9, 4, 6, 2, 8]

numbers.sort()

print(numbers)


# ------------------------------------------------------------
# Topic: Sorting Complexity Comparison
# ------------------------------------------------------------
print("\n========== Topic: Sorting Complexity ==========")

print("Bubble Sort")
print("Best Case    : O(n)")
print("Average Case : O(n²)")
print("Worst Case   : O(n²)")
print()

print("Selection Sort")
print("Best Case    : O(n²)")
print("Average Case : O(n²)")
print("Worst Case   : O(n²)")
print()

print("Insertion Sort")
print("Best Case    : O(n)")
print("Average Case : O(n²)")
print("Worst Case   : O(n²)")
print()

print("Merge Sort")
print("Best Case    : O(n log n)")
print("Average Case : O(n log n)")
print("Worst Case   : O(n log n)")
print()

print("Quick Sort")
print("Best Case    : O(n log n)")
print("Average Case : O(n log n)")
print("Worst Case   : O(n²)")


```

```python
# ============================================================
# PART 9 - FUNCTIONS IN PYTHON
# ============================================================

from functools import reduce

# ------------------------------------------------------------
# Topic: User Defined Function
# ------------------------------------------------------------
print("\n========== Topic: User Defined Function ==========")

def greet(name):
    return f"Hello, {name}!"

print(greet("Alice"))


# ------------------------------------------------------------
# Topic: Function with Required Arguments
# ------------------------------------------------------------
print("\n========== Topic: Required Arguments ==========")

def add(a, b):
    return a + b

print(add(3, 5))


# ------------------------------------------------------------
# Topic: Default Arguments
# ------------------------------------------------------------
print("\n========== Topic: Default Arguments ==========")

def greet_default(name="Guest"):
    return f"Hello, {name}!"

print(greet_default())
print(greet_default("John"))


# ------------------------------------------------------------
# Topic: Keyword Arguments
# ------------------------------------------------------------
print("\n========== Topic: Keyword Arguments ==========")

def student_info(name, age):
    return f"Name: {name}, Age: {age}"

print(student_info(age=20, name="Alice"))


# ------------------------------------------------------------
# Topic: Arbitrary Arguments (*args)
# ------------------------------------------------------------
print("\n========== Topic: *args ==========")

def sum_numbers(*nums):
    return sum(nums)

print(sum_numbers(1, 2, 3, 4, 5))


# ------------------------------------------------------------
# Topic: Arbitrary Keyword Arguments (**kwargs)
# ------------------------------------------------------------
print("\n========== Topic: **kwargs ==========")

def student_details(**info):
    return info

print(student_details(name="John", age=20, grade="A"))


# ------------------------------------------------------------
# Topic: Lambda Function
# ------------------------------------------------------------
print("\n========== Topic: Lambda Function ==========")

square = lambda x: x ** 2

print(square(5))


# ------------------------------------------------------------
# Topic: Lambda Function with Multiple Arguments
# ------------------------------------------------------------
print("\n========== Topic: Lambda with Multiple Arguments ==========")

add_lambda = lambda a, b: a + b

print(add_lambda(3, 7))


# ------------------------------------------------------------
# Topic: Return Statement
# ------------------------------------------------------------
print("\n========== Topic: Return Statement ==========")

def multiply(a, b):
    return a * b

result = multiply(4, 5)

print(result)


# ------------------------------------------------------------
# Topic: Local Variable
# ------------------------------------------------------------
print("\n========== Topic: Local Variable ==========")

def local_demo():
    x = 10
    print("Local Variable:", x)

local_demo()


# ------------------------------------------------------------
# Topic: Global Variable
# ------------------------------------------------------------
print("\n========== Topic: Global Variable ==========")

x = 20

def global_demo():
    print("Global Variable:", x)

global_demo()


# ------------------------------------------------------------
# Topic: global Keyword
# ------------------------------------------------------------
print("\n========== Topic: global Keyword ==========")

value = 10

def change_global():
    global value
    value = 50

change_global()

print(value)


# ------------------------------------------------------------
# Topic: Recursive Function (Factorial)
# ------------------------------------------------------------
print("\n========== Topic: Recursive Function ==========")

def factorial(n):

    if n == 1:
        return 1

    return n * factorial(n - 1)

print(factorial(5))


# ------------------------------------------------------------
# Topic: map() Function
# ------------------------------------------------------------
print("\n========== Topic: map() Function ==========")

nums = [1, 2, 3, 4]

squared = list(map(lambda x: x ** 2, nums))

print(squared)


# ------------------------------------------------------------
# Topic: filter() Function
# ------------------------------------------------------------
print("\n========== Topic: filter() Function ==========")

nums = [1, 2, 3, 4, 5, 6]

even = list(filter(lambda x: x % 2 == 0, nums))

print(even)


# ------------------------------------------------------------
# Topic: reduce() Function
# ------------------------------------------------------------
print("\n========== Topic: reduce() Function ==========")

nums = [1, 2, 3, 4]

product = reduce(lambda x, y: x * y, nums)

print(product)


# ------------------------------------------------------------
# Topic: Function Annotations (Type Hints)
# ------------------------------------------------------------
print("\n========== Topic: Function Annotations ==========")

def add_numbers(a: int, b: int) -> int:
    return a + b

print(add_numbers(2, 3))


```
